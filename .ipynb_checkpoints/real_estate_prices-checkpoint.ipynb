```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor

from sklearn.metrics import r2_score
from sklearn.metrics import mean_absolute_error
```

```python
house_price = sklearn.datasets.fetch_california_housing()
```

```python
data = pd.DataFrame(house_price.data, columns = house_price.feature_names)
data
```

```python
data['price'] = house_price['target']
```

```python
data.head()
```

## Create Model

```python
x = data.drop('price', axis = 1)
y = data.price
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = .2)
```

```python
model1 = XGBRegressor()
```

```python
model1.fit(x_train, y_train)
```

```python
pre = model1.predict(x_test)
pre
```

```python
metric1 = r2_score(y_test, pre)
metric1
```

```python
metric2 = mean_absolute_error(y_test, pre)
metric2
```
